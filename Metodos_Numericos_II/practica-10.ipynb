<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    <span style="color:#FF0000"><b>Ejercicio 1.</b></span> Al aplicar el método de Gauss con estrategia de pivot parcial al sistema anterior, se obtiene un sistema triangular superior equivalente, que posteriormente se resuelve mediante un proceso de remonte. Escribir dicho sistema triangular superior.
    </span></div>

```python
A = array([[-33, 36, -27, 18, -9], [-33, 51,-45, 30, -15], [ -19, 38, -45, 34, -17], [ -9, 18, -27, 27, -15],
           [ -3, 6, -9, 12, -9]])
print("A = ", A)

B=array([[120],[186],[162], [99],[36]])

#Gauss con pivote parcial pero muestra MA, MB
#MA = GAUSSA
#MB = GAUSSB
def mostrarGauss(A, B): #con pivote parcial
    m, n = shape(A)
    p, q = shape(B)
    if m != n or n != p or q < 1:
        return False, "Error gauss_pp: error en las dimensiones."
    if A.dtype == complex or B.dtype == complex:
        gaussA = array(A, dtype=complex) # a_1, a_2, a_3 ...
        gaussB = array(B, dtype=complex) # b_1, b_2, b_3 ...
    else:
        gaussA = array(A, dtype=float)
        gaussB = array(B, dtype=float)
        
    for k in range(n-1): #va de 0 a n-2  //Aquí empieza el buble principal del método de Gauss => busquemos el pívot
        
        print("\n\nIteración: k =",k+1)
        
        pos = argmax(abs(gaussA[k:, k])) #Elección del pívot (argmax = índice donde se encuentra el máximo de los elementos)
        ik = pos+k                       #Ya que necesitamos el a_kk como la fila 0 (argmax empieza a contar desde 0
        print("Posición del pivote: ik =",ik+1)
        
        if ik != k:     #Si el pivot no está en la posición 0, permutamos las filas (intercambiamos las fila ik-esima y la k-esima)                
            gaussA[[ik, k], :] = gaussA[[k, ik], :]
            gaussB[[ik, k], :] = gaussB[[k, ik], :]
        if abs(gaussA[k, k]) >= 1e-200:
            for i in range(k+1, n):
                gaussA[i, k] = gaussA[i, k]/gaussA[k, k]   #guarda los cocientes donde tenemos los ceros (el programa los ignora
                                                           #pensando que hay ceros, pero nosotros aprovechamos para guardas informació ahí)
                gaussA[i, k+1:] -= gaussA[i, k]*gaussA[k, k+1:]
                gaussB[i, :] -= gaussA[i, k]*gaussB[k, :]
#            print("\n Matriz A_k+1:\n",gaussA)
#            print("\n Matriz B_k+1:\n",gaussB)
                
#    print("\nMatriz triangualar MA:\n",triu(gaussA)) #Imprime la parte diagonal superior de la matriz (en el resto habíamos guardado
                                                      #información economizando, por lo que si la mostrásemos entera, no sería ceros)
#    print("\nSegundo miembro MB:\n",gaussB)
#
    MA=triu(gaussA)
    MB=gaussB
    exito, X = remonte(gaussA, gaussB)
    return exito, X, MA,MB


#######################

exito, X, MA, MB = mostrarGauss(A,B)
if(not exito):
    print("\n Error: no se pudo resolver el sistema")
else:
    print("\n Matriz triangular superior es MA =\n", MA)
    print("\n Vector MB=\n",MB)
    print("\n La solución tras el remonte es X=\n",X)
```

    <span style="color:#FF0000"><b>Ejercicio 1.</b></span> Dado el vector $X = \left( \begin{array}{c} 1 \\ 2 \\ 3 \\ 4 \end{array} \right) \in \mathbb{R}^4$, calcular la matriz de Householder $H(X) \in \mathcal{M}_4(\mathbb{R})$.

```python
# Ejercicio 1.
X = array([[1], [2], [3], [4]], dtype=float)
print("Vector: X = \n", X)
H = eye(4) - (2/(conjugada(X)@X))*(X@conjugada(X))
print("Matriz: H(X) = \n", H)
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    <span style="color:#FF0000"><b>Ejercicio 3.</b></span> Considera el sistema lineal $A\,X=B$, así como el sistema perturbado $A\,\bar{X}=\bar{B}$, donde
\[
A = \left( \begin{array}{rrrr} 10 & 1 & 4 & 0 \\ 1 & 10 & 5 & -1 \\ 4 & 5 & 10 & 7 \\ 0 & -1 & 7 & 9 \end{array} \right)\,, \quad B = \left( \begin{array}{r} 15 \\ 15 \\ 26 \\ 15 \end{array} \right) \quad \mbox{y} \quad \bar{B} = \left( \begin{array}{r} 16 \\ 16 \\ 25 \\ 16 \end{array} \right)\,.
\]
    Calcular $X$ y $\bar{X}$ (mediante el método de Gauss con estrategia de pívot parcial) y verificar la desigualdad
\[
\frac{\|\delta X\|_2}{\|X\|_2} \le \rm{cond}_2(A)\,\frac{\|\delta B\|_2}{\|B\|_2}\,.
\]
    </span></div>

```python
# Ejercicio 3.
A = array([[10,1,4,0], [1,10,5,-1], [4,5,10,7], [0,-1,7,9]])
B = array([[15], [15], [26], [15]])
Bmod = array([[16], [16], [25], [16]])

print("Matriz A = \n", A)
print("Matriz B = \n", B)
print("Matriz C = \n", Bmod)

exito, X = gauss_pp(A,B)
if exito:
    print("\nLa solución de AX = B es X = \n", X)

exito, Xmod = gauss_pp(A,Bmod)
if exito:
    print("\nLa solución de AX = Bmod es Xmod = \n", Xmod)
    
print("\nANÁLISIS DE ERRORES:")
errX = Xmod - X
condA = norma_mat(A,2)*norma_mat(inv(A),2)

print("Norma de X: ||X||_2 = ", norma_vec(X,2))
print("Norma de delta_X: ||delta_X||_2 = ", norma_vec(errX,2))

errB = Bmod - B
print("Norma de B: ||B||_2 = ", norma_vec(B,2))
print("Norma de delta_B: ||delta_B||_2 = ", norma_vec(errB,2))

print("Error relativo solución: ||delta_X||_2/||X||_2 = ", norma_vec(errX,2)/norma_vec(X,2))
print("Condicionamiento: cond_2(A) = ", condA)
print("Error relativo datos: ||delta_B||_2/||B||_2 = ", norma_vec(errB,2)/norma_vec(B,2))

exito = norma_vec(errX,2)/norma_vec(X,2) <= condA * norma_vec(errB,2)/norma_vec(B,2)
print("Verificación del teorema: ", exito)
print("Primer término de la desigualdad: ", norma_vec(errX,2)/norma_vec(X,2))
print("Segundo término de la desigualdad: ", condA * norma_vec(errB,2)/norma_vec(B,2))
```

<div align="left"><span style="font-family: Arial; color:#000000; font-size: medium">
    <span style="color:#FF0000"><b>Ejercicio 4.</b></span> Calcular $\rm{cond}_p(A_2)$, $p=1,2,\infty$, donde $A_2$ es la matriz triangular superior resultante de aplicar el método de Gauss al sistema
\[
\left\{ \begin{array}{rcrcr} 10^{-4}\,x_1 & + & x_2 & = & 1\,, \\ x_1 & + & x_2 & = & 2\,. \end{array} \right.
\]
    Comparar los resultados cuando no se utiliza ninguna estrategia de pivot y cuando se utiliza una estrategia de pivot parcial.
    <br>
    Cambiar el elemento $(1,1)$ de la matriz de coeficientes por $10^{-16}$ y analizar el resultado.
    </span></div>

```python
def gauss_pp_DevTriangSup(A, B):
    m, n = shape(A)
    p, q = shape(B)
    if m != n or n != p or q < 1:
        return False, "Error gauss_pp_DevTriangSup: error en las dimensiones."
    if A.dtype == complex or B.dtype == complex:
        gaussA = array(A, dtype=complex)
        gaussB = array(B, dtype=complex)
    else:
        gaussA = array(A, dtype=float)
        gaussB = array(B, dtype=float)
    for k in range(n-1):
        pos = argmax(abs(gaussA[k:, k])) # Elección del pívot (argmax da la posicion!! del maximo valor de los elementos)
        ik = pos+k # Posicion absoluta (no relativa al tamaño k-ésimo!!)
        if ik != k: # Si pivot no está en posición natural, permutamos las filas
            gaussA[[ik, k], :] = gaussA[[k, ik], :]
            gaussB[[ik, k], :] = gaussB[[k, ik], :]
        if abs(gaussA[k, k]) >= 1e-200:
            for i in range(k+1, n):
                gaussA[i, k] = gaussA[i, k]/gaussA[k, k] 
                gaussA[i, k+1:] -= gaussA[i, k]*gaussA[k, k+1:]
                gaussB[i, :] -= gaussA[i, k]*gaussB[k, :]
    exito, X = remonte(gaussA, gaussB)
    return exito, X, triu(gaussA)
```

```python
def metodo_lu_DevTriangSup(A, B):
    m, n = shape(A)
    p, q = shape(B)
    if m!= n or n != p or q < 1:
        return False, "Error metodo_lu: Tamaños erróneos"
    exito,lu = facto_lu(A)
    if exito:
        ex,Y = descenso1(lu,B)
        exito1,X = remonte(lu,Y)
        if exito1:
            return True,X, triu(lu)
        else:
            return False, "Error metodo_lu: factorización LU no inversible"
    else:
        return False, "Error metodo_lu: error en la resolucion"
```

```python
# Ejercicio 4.
A = array([[1e-4,1],[1,1]])
B = array([[1],[2]])
print("Matriz A=\n",A)
print("Array B=\n",B)

#NINGUNA ESTRATEGIA DE PIVOT = LU

(exito,X,triangSup) = gauss_pp_DevTriangSup(A,B)

print("\nLa matriz triangular superior del método de Gauss es \n", triangSup)
print("\nLa solución del sistema por método de Gauss con pivote es X=\n",X)

print("\ncond1=",condicionamiento(triangSup,1))
print("cond2=",condicionamiento(triangSup,2))
print("cond(inf)=",condicionamiento(triangSup,inf))

(exito,X,triangSup) = metodo_lu_DevTriangSup(A,B)
print(X)

print("\nLa matriz triangular superior del método LU (Gauss sin pivote) es \n", triangSup)
print("\nLa solución del sistema por método LU (Gauss sin pivote) es X=\n",X)

print("\ncond1=",condicionamiento(triangSup,1))
print("cond2=",condicionamiento(triangSup,2))
print("cond(inf)=",condicionamiento(triangSup,inf))


#Cuando cambiamos el a11 por 1e-16 en vez de 1e-4, ocurre que el método LU no da la solución correcta
# y el condicionamiento se dispara (este es el problema de que no se resuelva bien el sistema)
```
